# Sanity Check

This notebook simply tests the loading of transformer lens and the GPT-2 small model. We also generate a small set of toy inputs to determine how we would like to structure our prompts to assess the greater than capability.

Some key considerations:
- The max token length of the output may influence the accuracy of the model outputs
- Structure of prompt may influence accuracy of model outputs

```python
"""Load GPT-2 and Simple Forward Pass"""
from transformer_lens import HookedTransformer
import torch

# Load model
model = HookedTransformer.from_pretrained("gpt2-small")

# Create sample input for forward pass
text = "The cat sat on the"
output = model.generate(text, max_new_tokens=10)

# Get the logits
logits = model(text)

print(logits)
```

## Toy Samples

Check what types of prompts GPT-2 performs well on and struggles on. It should be apparent that GPT-2 struggles haevily on math notation since it was trained on natural language. Interestingly it also does not do particularly well on multiple choice.

```python
"""Generate simple test prompts to see model behavior"""

# Sample input prompt 1
prompt1 = "1728 is less than 1748. 1650 is less than 1800. 1920 is less than"

# Get output and print
output = model.generate(prompt1, max_new_tokens=5)

print("Sample input 1: \n")
print(output)
print("\n\n")

# Sample input prompt 2
prompt2 = "Which year comes later: (A) 1728 or (B) 1748? Answer:"

# Get output and print
output = model.generate(prompt2, max_new_tokens=5)

print("Sample input 2: \n")
print(output)
print("\n\n")

# Sample input prompt 3
prompt3 = "1728, then 1748. 1650, then 1800. 1920, then"
# Get output and print
output = model.generate(prompt3, max_new_tokens=5)

print("Sample input 2: \n")
print(output)
print("\n\n")
```

```python
import random
```

```python
"""Create many natural language samples"""

for i in range(20):
    # Generate bounds
    small = random.randint(0, 100)
    large = random.randint(100, 200)
    compare = random.randint(0, 200)

    # Generate prompt
    prompt = f"{large} is greater than {small}. {compare} is greater than"

    # Get output
    output = model.generate(prompt, max_new_tokens=2)

    # Print
    print(output)
```
